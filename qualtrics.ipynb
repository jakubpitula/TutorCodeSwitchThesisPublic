```python
# Get the data from the surveys file and format it for qualtrics

with open("surveys_new.json", "r") as f:
    surveys_final = json.load(f)
```

```python
formatted_surveys = []
for survey in surveys_final:
    formatted_survey = []
    for topic, languages in survey.items():
        for language, samples in languages.items():
            for sample in samples:
                context = sample["Context"]
                formatted_context = ""
                for turn in context:
                    speaker = "TUTOR" if turn["Speaker"] == "INV" else "STUDENT"
                    formatted_context += f"{speaker}: {turn['Turn']}<br>\n"

                formatted_survey.append(
                    {
                        "context": formatted_context,
                        "tutor": sample["Tutor"],
                        "topic": topic,
                        "language": language,
                    }
                )
    random.shuffle(formatted_survey)
    formatted_surveys.append(formatted_survey)
```

```python
formatted_survey_teacher = []
for topic, languages in teacher_survey_final.items():
    for language, samples in languages.items():
        for sample in samples:
            print(sample)
            context = sample["Context"]
            formatted_context = ""
            for turn in context:
                speaker = "TUTOR" if turn["Speaker"] == "INV" else "STUDENT"
                formatted_context += f"{speaker}: {turn['Turn']}<br>\n"

            formatted_survey_teacher.append(
                {
                    "context": formatted_context,
                    "tutor": sample["Tutor"],
                    "topic": topic,
                    "language": language,
                }
            )
random.shuffle(formatted_survey_teacher)
```
